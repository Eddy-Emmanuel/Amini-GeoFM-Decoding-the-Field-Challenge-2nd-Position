```python
BANDS = ['B2', 'B3', 'B4', 'B8', 'B11', 'B12', 'B5', 'B6', 'B7', 'B8A']
BAND_NAMES = ['blue', 'green', 'red', 'nir', 'swir16', 'swir22', 'rededge1', 'rededge2', 'rededge3', 'nir08']

START_DATE = '2023-01-01'
END_DATE = '2023-12-31'
TIME_STEP_DAYS = 7

# --- 3. GEE Functions ---
def maskS2clouds(image):
    """Masks clouds in a Sentinel-2 image using the QA band."""
    qa = image.select('QA60')
    cloudBitMask = 1 << 10
    cirrusBitMask = 1 << 11
    mask = qa.bitwiseAnd(cloudBitMask).eq(0).And(qa.bitwiseAnd(cirrusBitMask).eq(0))
    return image.updateMask(mask).divide(10000)

def process_request(task):
    """
    Worker function for multiprocessing. Processes one polygon for one date range.
    """

    poly_index, poly_info, start_date = task
    polygon = ee.Geometry.Polygon(poly_info['coordinates'])
    end_date = start_date + pd.DateOffset(days=TIME_STEP_DAYS)

    try:
        s2_collection = (
            ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
            .filterDate(start_date.strftime('%Y-%m-%d'), end_date.strftime('%Y-%m-%d'))
            .filterBounds(polygon)
            .map(maskS2clouds)
            .select(BANDS, BAND_NAMES)
        )
        composite = s2_collection.median()
        stats = composite.reduceRegion(
            reducer=ee.Reducer.mean(),
            geometry=polygon,
            scale=10,
            maxPixels=1e9
        ).getInfo()
    except Exception as e:
        # If any GEE error occurs, return an empty dictionary for stats
        print(f"Error processing polygon {poly_index+1} for date {start_date.strftime('%Y-%m-%d')}: {e}")
        stats = {}

    row_data = {
        'polygon_id': poly_index + 1,
        'date': start_date.strftime('%Y-%m-%d'),
    }
    for band_name in BAND_NAMES:
        row_data[band_name] = stats.get(band_name)

    return row_data

# --- 4. Data Extraction using Multiprocessing ---
polygons = geometry.geometries()
date_ranges = pd.date_range(start=START_DATE, end=END_DATE, freq=f'{TIME_STEP_DAYS}D')
polygon_list = polygons.getInfo()

# Create a flat list of all tasks to be processed
tasks = list(itertools.product(enumerate(polygon_list), date_ranges))
# Unpack the nested tuple from itertools.product
tasks = [(i, poly_info, start_date) for ((i, poly_info), start_date) in tasks]

print(f"Created {len(tasks)} tasks to process using {cpu_count()} CPU cores...")

all_results = []
# Use a multiprocessing Pool to execute tasks in parallel
with Pool(processes=cpu_count()) as pool:
    # tqdm will track the progress as results are yielded from the pool
    with tqdm(total=len(tasks), desc="Extracting GEE Data") as pbar:
        for result in pool.imap_unordered(process_request, tasks):
            all_results.append(result)
            pbar.update()

print("Data extraction complete.")

# --- 5. Create DataFrame and Display Results ---
df = pd.DataFrame(all_results)
df = df.sort_values(by=['polygon_id', 'date']).reset_index(drop=True)

# Convert band columns to numeric, coercing errors to NaN
for band_name in BAND_NAMES:
    df[band_name] = pd.to_numeric(df[band_name], errors='coerce')

print("\n--- Head of the Downloaded Data ---")
print(df.head())

print("\n--- Statistics per Band ---")
print(df[BAND_NAMES].describe())

# --- 6. Save Data to Kaggle Output ---
output_path = '/kaggle/working/Palm_Plantation_CI.csv'
df.to_csv(output_path, index=False)
print(f"\n✅ Data successfully saved to {output_path}")
```

```python
BANDS = ['B2', 'B3', 'B4', 'B8', 'B11', 'B12', 'B5', 'B6', 'B7', 'B8A']
BAND_NAMES = ['blue', 'green', 'red', 'nir', 'swir16', 'swir22', 'rededge1', 'rededge2', 'rededge3', 'nir08']

START_DATE = '2023-01-01'
END_DATE = '2023-12-31'
TIME_STEP_DAYS = 7

# --- 3. GEE Functions ---
def maskS2clouds(image):
    """Masks clouds in a Sentinel-2 image using the QA band."""
    qa = image.select('QA60')
    cloudBitMask = 1 << 10
    cirrusBitMask = 1 << 11
    mask = qa.bitwiseAnd(cloudBitMask).eq(0).And(qa.bitwiseAnd(cirrusBitMask).eq(0))
    return image.updateMask(mask).divide(10000)

def process_request(task):
    """
    Worker function for multiprocessing. Processes one polygon for one date range.
    """

    poly_index, poly_info, start_date = task
    polygon = ee.Geometry.Polygon(poly_info['coordinates'])
    end_date = start_date + pd.DateOffset(days=TIME_STEP_DAYS)

    try:
        s2_collection = (
            ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
            .filterDate(start_date.strftime('%Y-%m-%d'), end_date.strftime('%Y-%m-%d'))
            .filterBounds(polygon)
            .map(maskS2clouds)
            .select(BANDS, BAND_NAMES)
        )
        composite = s2_collection.median()
        stats = composite.reduceRegion(
            reducer=ee.Reducer.mean(),
            geometry=polygon,
            scale=10,
            maxPixels=1e9
        ).getInfo()
    except Exception as e:
        # If any GEE error occurs, return an empty dictionary for stats
        print(f"Error processing polygon {poly_index+1} for date {start_date.strftime('%Y-%m-%d')}: {e}")
        stats = {}

    row_data = {
        'polygon_id': poly_index + 1,
        'date': start_date.strftime('%Y-%m-%d'),
    }
    for band_name in BAND_NAMES:
        row_data[band_name] = stats.get(band_name)

    return row_data

# --- 4. Data Extraction using Multiprocessing ---
polygons = geometry.geometries()
date_ranges = pd.date_range(start=START_DATE, end=END_DATE, freq=f'{TIME_STEP_DAYS}D')
polygon_list = polygons.getInfo()

# Create a flat list of all tasks to be processed
tasks = list(itertools.product(enumerate(polygon_list), date_ranges))
# Unpack the nested tuple from itertools.product
tasks = [(i, poly_info, start_date) for ((i, poly_info), start_date) in tasks]

print(f"Created {len(tasks)} tasks to process using {cpu_count()} CPU cores...")

all_results = []
# Use a multiprocessing Pool to execute tasks in parallel
with Pool(processes=cpu_count()) as pool:
    # tqdm will track the progress as results are yielded from the pool
    with tqdm(total=len(tasks), desc="Extracting GEE Data") as pbar:
        for result in pool.imap_unordered(process_request, tasks):
            all_results.append(result)
            pbar.update()

print("Data extraction complete.")

# --- 5. Create DataFrame and Display Results ---
df = pd.DataFrame(all_results)
df = df.sort_values(by=['polygon_id', 'date']).reset_index(drop=True)

# Convert band columns to numeric, coercing errors to NaN
for band_name in BAND_NAMES:
    df[band_name] = pd.to_numeric(df[band_name], errors='coerce')

print("\n--- Head of the Downloaded Data ---")
print(df.head())

print("\n--- Statistics per Band ---")
print(df[BAND_NAMES].describe())

# --- 6. Save Data to Kaggle Output ---
output_path = '/kaggle/working/S2_7day_Rubber.csv'
df.to_csv(output_path, index=False)
print(f"\n✅ Data successfully saved to {output_path}")
```

```python
BANDS = ['B2', 'B3', 'B4', 'B8', 'B11', 'B12', 'B5', 'B6', 'B7', 'B8A']
BAND_NAMES = ['blue', 'green', 'red', 'nir', 'swir16', 'swir22', 'rededge1', 'rededge2', 'rededge3', 'nir08']

START_DATE = '2020-01-01'
END_DATE = '2023-12-31'
TIME_STEP_DAYS = 7

# --- 3. GEE Functions ---
def maskS2clouds(image):
    """Masks clouds in a Sentinel-2 image using the QA band."""
    qa = image.select('QA60')
    cloudBitMask = 1 << 10
    cirrusBitMask = 1 << 11
    mask = qa.bitwiseAnd(cloudBitMask).eq(0).And(qa.bitwiseAnd(cirrusBitMask).eq(0))
    return image.updateMask(mask).divide(10000)

def process_request(task):
    """
    Worker function for multiprocessing. Processes one polygon for one date range.
    """

    poly_index, poly_info, start_date = task
    polygon = ee.Geometry.Polygon(poly_info['coordinates'])
    end_date = start_date + pd.DateOffset(days=TIME_STEP_DAYS)

    try:
        s2_collection = (
            ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
            .filterDate(start_date.strftime('%Y-%m-%d'), end_date.strftime('%Y-%m-%d'))
            .filterBounds(polygon)
            .map(maskS2clouds)
            .select(BANDS, BAND_NAMES)
        )
        composite = s2_collection.median()
        stats = composite.reduceRegion(
            reducer=ee.Reducer.mean(),
            geometry=polygon,
            scale=10,
            maxPixels=1e9
        ).getInfo()
    except Exception as e:
        # If any GEE error occurs, return an empty dictionary for stats
        print(f"Error processing polygon {poly_index+1} for date {start_date.strftime('%Y-%m-%d')}: {e}")
        stats = {}

    row_data = {
        'polygon_id': poly_index + 1,
        'date': start_date.strftime('%Y-%m-%d'),
    }
    for band_name in BAND_NAMES:
        row_data[band_name] = stats.get(band_name)

    return row_data

# --- 4. Data Extraction using Multiprocessing ---
polygons = geometry.geometries()
date_ranges = pd.date_range(start=START_DATE, end=END_DATE, freq=f'{TIME_STEP_DAYS}D')
polygon_list = polygons.getInfo()

# Create a flat list of all tasks to be processed
tasks = list(itertools.product(enumerate(polygon_list), date_ranges))
# Unpack the nested tuple from itertools.product
tasks = [(i, poly_info, start_date) for ((i, poly_info), start_date) in tasks]

print(f"Created {len(tasks)} tasks to process using {cpu_count()} CPU cores...")

all_results = []
# Use a multiprocessing Pool to execute tasks in parallel
with Pool(processes=cpu_count()) as pool:
    # tqdm will track the progress as results are yielded from the pool
    with tqdm(total=len(tasks), desc="Extracting GEE Data") as pbar:
        for result in pool.imap_unordered(process_request, tasks):
            all_results.append(result)
            pbar.update()

print("Data extraction complete.")

# --- 5. Create DataFrame and Display Results ---
df = pd.DataFrame(all_results)
df = df.sort_values(by=['polygon_id', 'date']).reset_index(drop=True)

# Convert band columns to numeric, coercing errors to NaN
for band_name in BAND_NAMES:
    df[band_name] = pd.to_numeric(df[band_name], errors='coerce')

print("\n--- Head of the Downloaded Data ---")
print(df.head())

print("\n--- Statistics per Band ---")
print(df[BAND_NAMES].describe())

# --- 6. Save Data to Kaggle Output ---
output_path = '/kaggle/working/S2_7day_Palm.csv'
df.to_csv(output_path, index=False)
print(f"\n✅ Data successfully saved to {output_path}")
```
